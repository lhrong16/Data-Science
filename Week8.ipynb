## **0. Data initialization**


```python
import numpy as np
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')
df_salary = pd.read_csv('/content/drive/MyDrive/Excel files/salary_cleaned.csv') #load the csv dataset into a 'dataframe' df
df_xtra_info = pd.read_csv('/content/drive/MyDrive/Excel files/xtra_info (1).csv') #load the csv dataset into a 'dataframe' df
df_master = pd.concat([df_salary,df_xtra_info],axis=1)
print(len(df_master))
df_master.head()
```

## **1. The higher the education level, the higher the Salary**

```python
df_master.plot.hist(column='Salary', by='Education level', figsize=(7, 30))
```

## **2. The more experience a person is, the higher the Salary**

```python
df_master.plot.hist(column='Salary', by='Years of experience in current job', figsize=(7, 30))
```
